**Предоставитель данных:** Kaggle <br>
**Наименование:** данные о пассажирах на борту Титаника с разметкой целевого признака <br>
**Формат файла с данными:** csv <br>
**Путь к файлу:** /datasets/raw/train.csv

**Изначальная структура данных (без обработки):**

- **titanic**
- `PassengerId` — уникальный номер пассажира;
- `Survived` — выживание в катастрофе 0 = Нет, 1 = Да;
- `Pclass` — показатель социально-экономического статуса:
    * 1-е место = высший;
    * 2-е место = средний;
    * 3-е место = низший.
- `Name` — имя пассажира;
- `Sex` — пол;
- `Age` — Возраст в годах (возраст указывается дробным числом, если он меньше 1, если возраст указан приблизительно, то в виде xx.5);
- `SibSp` — Количество братьев и сестер / супругов на борту «Титаника»:
    * Sibling = брат, сестра, сводный брат, сводная сестра;
    * Spouse = муж, жена (любовницы и невесты остались без внимания).
- `Parch` — родителей / детей на борту «Титаника»:
    * Parent = мать, отец;
    * Child = доч, сын, сводная дочь, сводный сын;
    * Некоторые дети путешествовали исключительно с няней, поэтому для них parch=0.
- `Ticket` — номер билета;
- `Fare` — стоимость проезда для пассажиров;
- `Cabin` — номер каюты;
- `Embarked` - место посадки (порт):
- C = Cherbourg;
- Q = Queenstown;
- S = Southampton.
<br><br>
